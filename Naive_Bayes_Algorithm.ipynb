```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("play_tennis.csv")
```

```python
data.head()
```

```python
# Drop the day coloums because there is no meaning of that coloumn
data.drop(columns=['day'] , inplace=True) # we use inplace = True so that it modify the existing data
```

```python
# Now if we are given an problem that outlook= sunny, temp= hot, humidity= High, wind= weak
# and we have to predict whether there will be play on that day or not so by using naive bayes we can solve this as:
#
# P(Yes| sunny, hot , high, weak) will be written as : P(sunny|Yes) * P(hot|Yes) * P(High|Yes) * P(weak|Yes) * P(Yes)
# The denominator term is same in both case so we ignore it
# similarly
# P(No| sunny, hot , high , weak) and this can be written as: P(sunny|No) * P(hot|No) * P(High|No) * P(weak|No) * P(No)
```

```python
# Now how naive solve this is by creating a lookup table where all the probabilities are stored and if any problem given it can solve by calculating
# Now we will create a lookup for naive bayes
```

```python
# First we will calculate the probability of yes and no and it can be calculates as:

data['play'].value_counts()
```

```python
P_Yes = 9/14
P_No = 5/14

print(P_Yes)
print(P_No)
```

```python
# Now  we will calculate the probabilities of all the coloums
# 1st is Outlook coloumn

pd.crosstab(data['outlook'], data['play'])
```

```python
P_Overcast_Yes = 4/9
P_Rain_Yes = 3/9
P_Sunny_Yes = 2/9

P_Overcast_No = 0/5
P_Rain_No = 2/5
P_Sunny_No = 3/5
```

```python
# Now 2nd coloum which is Temp

pd.crosstab(data['temp'], data['play'])
```

```python
P_Cool_Yes = 3/9
P_Hot_Yes = 2/9
P_Mild_Yes = 4/9

P_Cool_No = 1/5
P_Hot_No = 2/5
P_Mild_No = 2/5
```

```python
# Now 3rd coloum which is Humidity

pd.crosstab(data['humidity'], data['play'])
```

```python
P_High_Yes = 3/9
P_Normal_Yes = 6/9

P_High_No = 4/5
P_Normal_No = 1/5
```

```python
# Now 4th coloum which is Wind
pd.crosstab(data['wind'], data['play'])
```

```python
P_Strong_Yes = 3/9
P_Weak_Yes = 6/9

P_Srong_No = 3/5
P_Weak_No = 2/5
```

```python
# Now we will solve our given problem
# outlook= sunny, temp= hot, humidity= High, wind= weak

Probability_of_Yes = P_Yes * P_Sunny_Yes * P_Hot_Yes * P_High_Yes * P_Weak_Yes
#Similarly
Probability_of_No = P_No * P_Sunny_No * P_Hot_No * P_High_No * P_Weak_No

# Now check both probability and which one is greater in our answer
print(Probability_of_Yes)
print(Probability_of_No)
```
