```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
# we use this class to generate the dataset
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=20)
```

```python
# Plot the data to see how it is looking
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1], c=y , cmap='winter' , s=100)
# c is color and here we pass y which contai 0 and 1 and and the matplot lib automatically assign 1 to blue and 0 to green
#  cmap is the color map and we chose winter which is a built-in Matplotlib palette that shifts from vibrant blue to green
plt.colorbar(label='Target Classes (y)')
```

```python
# Now we will create our perception class.
def Perception(X,y):

  # first we will insert a coloumn for easy vector multiplication

  X = np.insert(X, 0, 1 , axis=1) # here we chose axis as 1 which means to add a colomn and if we chose axis=0 the row will be added
  weights = np.ones(X.shape[1]) # these are the coef_  for 1 row
  lr = 0.1

  for i in range(1000):
    j = np.random.randint(X.shape[0])
    y_hat = sigmoid(np.dot(X[j] , weights))
    weights = weights + lr*(y[j] - y_hat)*X[j]

  return weights[0], weights[1:]
# here we take the first coloumn of weights as intercept and other are coef_
```

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))

  # As we know in our logic the model will predict 1 if the dot product is greater than 1 and it will return 0 if the dot product is less than 0
```

```python
intercept_, coef_ = Perception(X,y)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.ylim(-3,2)
```

```python
from sklearn.linear_model import LogisticRegression
lor = LogisticRegression()
lor.fit(X,y)
```

```python
m = -(lor.coef_[0][0]/lor.coef_[0][1])
b = -(lor.intercept_/lor.coef_[0][1])
```
