```python
url = "https://jsearch.p.rapidapi.com/search-v2"

headers = {
	"x-rapidapi-key": "1f1e2bec18mshb80e2119ca485c4p14da70jsn794339e8db15",
	"x-rapidapi-host": "jsearch.p.rapidapi.com",
	"Content-Type": "application/json"
}

queries = [
    "Machine Learning Engineer in United States",
    "Data Scientist in United States",
    "AI Engineer in United States",
    "Data Analyst in United States",
    "Deep Learning Engineer in United States",
    "MLOps Engineer in United States",
]

all_jobs = []

```

```python
for q in queries:
  print(f"Fetching: {q}...")

  for page in range(1, 3):
    querystring = {
        "query": q,
        "page": str(page),
        "num_pages": "1",
        "date_posted": "all",
    }

    response = requests.get(url, headers=headers, params=querystring)

    if response.status_code == 200:
      res_json = response.json()
      # Access data -> jobs list
      jobs_list = res_json.get("data", {}).get("jobs", [])

      for job in jobs_list:
        all_jobs.append({
            "search_query": q,
            "job_title": job.get("job_title"),
            "employer_name": job.get("employer_name"),
            "job_employment_type": job.get("job_employment_type"),
            "job_city": job.get("job_city"),
            "job_state": job.get("job_state"),
            "job_country": job.get("job_country"),
            "job_is_remote": job.get("job_is_remote"),
            "min_salary": job.get("job_min_salary"),
            "max_salary": job.get("job_max_salary"),
            "salary_currency": job.get("job_salary_currency"),
            "salary_period": job.get("job_salary_period"),
            "posted_date": job.get("job_posted_at_datetime_utc"),
            "job_apply_link": job.get("job_apply_link"),
            "job_description": job.get("job_description"),
        })

    time.sleep(1)

print(f"\nFinished fetching! Total records collected: {len(all_jobs)}")
```

# Building a Tech Job Dataset from an API

A simple, beginner-friendly guide explaining what we did in **Day 17 of 100 Days of Machine Learning**, why we did it, and the lessons learned along the way.

---

## What Was the Goal?

In Day 17, Nitish explained how machine learning engineers don't always get clean, pre-made CSV files. Often, you have to collect your own data from web APIs.

Our mission was:
1. Learn how to talk to APIs using Python.
2. Fix real-world errors that happen when extracting web data.
3. Build a real, useful dataset from scratch that can be uploaded to Kaggle for data analysis or NLP projects.

---

## What is an API and What is RapidAPI?

* **API (Application Programming Interface):** A digital bridge that lets Python ask a website for raw information and get back clean text data instead of a visual webpage.
* **RapidAPI:** A marketplace for APIs. Instead of creating separate accounts on dozens of different websites, RapidAPI gives you **one central API key** that works across thousands of different services.
* **Where our data actually came from:** RapidAPI doesn't create the jobs data. It forwarded our requests to **JSearch**, an engine that indexes live job boards across the internet (like LinkedIn, Indeed, and Glassdoor).

---

## The Journey & Problems We Solved

Building datasets from APIs rarely works on the first try. Here is what happened step-by-step:

### Phase 1: The Shazam Experiment
* **The Goal:** We initially tried the Shazam API to get song suggestions and chart lists.
* **The Hurdle:** The endpoints either returned tiny lists (just 3 search suggestions) or sent empty `204 No Content` responses because the free endpoints were limited or changed.
* **The Lesson:** Always test an API endpoint with a single request first to see if it actually returns the rich, multi-row data you want before writing loops.

### Phase 2: Finding a Useful Dataset Idea (JSearch)
* Rather than gathering generic movie or song lists that already flood Kaggle, we switched to **live AI and Machine Learning job listings in the United States**.
* This type of data is valuable because it contains real job descriptions, required skills, company names, and employment types.

### Phase 3: Fixing Common Python & API Errors
1. **JSON Decode Error:** This happened when we called `.json()` on an HTML homepage or an empty response. A computer cannot turn a webpage design into a Python dictionary.
2. **List Indexing Error:** Trying to filter columns using list syntax failed because Python lists only take numbers (indexes), not column names. Column filtering must happen **after** creating the Pandas DataFrame.
3. **Attribute Error (`'str' object has no attribute 'get'`):** The new version of the API nested the list of jobs inside `data -> jobs`. We accidentally looped through the dictionary keys instead of the job list itself. Targeting the nested `jobs` list solved it immediately.

---

## How the Final Pipeline Works

The entire workflow follows four logical steps:

1. **Setup & Authentication:**
   * We gave Python the endpoint URL and our RapidAPI security key so the server knew who was asking for the data.
   * We picked 6 high-demand tech roles: *Machine Learning Engineer, Data Scientist, AI Engineer, Data Analyst, Deep Learning Engineer,* and *MLOps Engineer*.

2. **The Automated Harvesting Loop:**
   * Python cycled through each job title one by one.
   * For each title, it visited multiple pages.
   * We added a 1-second pause between requests so we wouldn't get blocked or banned by the server for requesting data too fast.

3. **Data Extraction:**
   * From every job post returned, we picked out only the cleanest details:
     * Job Title
     * Company / Employer Name
     * Employment Type (Full-time, Contract, etc.)
     * Location (State and Country)
     * Remote Status
     * Date Posted
     * Direct Apply Link
     * Full Job Description

4. **Cleaning and Saving:**
   * All collected job listings were placed into a structured table using **Pandas**.
   * We checked for duplicate jobs (since different search terms might return the same opening) and removed them.
   * Finally, we exported the table into a CSV file ready for analysis.
